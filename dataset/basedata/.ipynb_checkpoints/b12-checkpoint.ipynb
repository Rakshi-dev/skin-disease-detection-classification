```python
# Import libraries and dataset
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras import models, layers
from tensorflow.keras.layers import Dense
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
IMAGE_SIZE = 256
BATCH_SIZE = 32
CHANNELS = 3
EPOCHS = 50
```

```python
training_dataset = tf.keras.preprocessing.image_dataset_from_directory(
   "C:\\Users\\admin\\Desktop\\dataset\\basedata\\training",
    shuffle=True,
    image_size = (256,256),
    batch_size = 32
)

testing_dataset = tf.keras.preprocessing.image_dataset_from_directory(
   "C:\\Users\\admin\\Desktop\\dataset\\basedata\\testing",
    shuffle=True,
    image_size = (256,256),
    batch_size = 32
)

validation_dataset = tf.keras.preprocessing.image_dataset_from_directory(
   "C:\\Users\\admin\\Desktop\\dataset\\basedata\\validation",
    shuffle=True,
    image_size = (256,256),
    batch_size = 32
)
```

```python
class_names = training_dataset.class_names
class_names
```

```python
for image_batch, label_batch in training_dataset.take(1):
    print(image_batch[0].shape)
```

```python
# Show an example of loaded images
plt.figure(figsize=(10,10))
for image_batch, label_batch in training_dataset.take(1):
    for i in range(12):
        plt.subplot(3,4,i+1)
        plt.imshow(image_batch[i].numpy().astype("uint8"))
        plt.title(class_names[label_batch[i]])
        plt.axis("off")
```

```python
train_set = training_dataset.cache().shuffle(50).prefetch(buffer_size=tf.data.AUTOTUNE)
valid_set = validation_dataset.cache().shuffle(50).prefetch(buffer_size=tf.data.AUTOTUNE)
test_set = testing_dataset.cache().shuffle(50).prefetch(buffer_size=tf.data.AUTOTUNE)
```

```python
train_size = 0.8
train_set = training_dataset.take(int(train_size*len(training_dataset)))
len(train_set)
```

```python
# Data Partitioning

val_test_set = training_dataset.skip(int(train_size*len( training_dataset)))
val_size = 0.1
val_set = val_test_set.take(int(val_size*len( training_dataset)))
test_set = val_test_set.skip(int(val_size*len( training_dataset)))
print("Validation set length:", len(val_set))
print("Test set length:", len(test_set))
```

```python
resize_scale = tf.keras.Sequential([
    layers.Resizing(256, 256),
    layers.Rescaling(1.0/255)
])
```

```python
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.2)
])

# Example usage with an input image
input_image = tf.random.uniform(shape=[1, 256, 256, 3])  # Example image tensor
augmented_image = data_augmentation(input_image)

print(augmented_image.shape)
```

```python
# Model Implementaion

input_shape = (32, 256, 256, 3)
n_classes = 3

# Convolutional NN model
model = models.Sequential([
    resize_scale,
    data_augmentation,
    layers.Conv2D(32, (3,3), activation = 'relu', input_shape = input_shape),
    layers.MaxPool2D((2,2)),
    layers.Conv2D(64, (3,3), activation = 'relu'),
    layers.MaxPool2D((2,2)),
    layers.Conv2D(64, (3,3), activation = 'relu'),
    layers.MaxPool2D((2,2)),
    layers.Conv2D(64, (3,3), activation = 'relu'),
    layers.MaxPool2D((2,2)),
    layers.Conv2D(64, (3,3), activation = 'relu'),
    layers.MaxPool2D((2,2)),
    layers.Conv2D(64, (3,3), activation = 'relu'),
    layers.MaxPool2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation = 'relu'),
    layers.Dense(n_classes, activation = 'softmax')
])

model.build(input_shape = input_shape)
```

```python
model.summary()
```

```python
#compile the model
model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False),
    metrics=['accuracy']
)
```

```python
# Model training with data
history = model.fit(
    train_set,
    epochs = 50,
    batch_size = 32,
    verbose = 1,
    validation_data =validation_dataset
)
```

```python
score = model.evaluate(test_set)
```

```python
train_acc = history.history['accuracy']
val_acc = history.history['val_accuracy']

plt.figure(figsize=(10,8))
plt.plot(range(50), train_acc, label = 'Training accuracy')
plt.plot(range(50), val_acc, label = 'Validation accuracy')
plt.legend(loc = 'lower right')
plt.title('Training and validation accuracy')
```

```python
train_loss = history.history['loss']
val_loss = history.history['val_loss']

plt.figure(figsize=(10,8))
plt.plot(range(50), train_loss, label = 'Training loss')
plt.plot(range(50), val_loss, label = 'Validation loss')
plt.legend(loc = 'upper right')
plt.title('Training and validation loss')
```

```python
# Model Prediction

for image_batch, label_batch in testing_dataset.take(1):
    
    first_image = image_batch[0].numpy().astype('uint8')
    first_label = label_batch[0]
    
    print("First image to predict")
    plt.imshow(first_image)
    print("First image's true label: ", class_names[first_label])
    
    batch_prediction = model.predict(image_batch)
    print("First image's predicted label: ", class_names[np.argmax(batch_prediction[0])])
```

```python
# Function for prediction and confidence
def predict_confidence(model, image):
    img_array = tf.keras.preprocessing.image.img_to_array(image.numpy())
    img_array = tf.expand_dims(img_array, 0)
    
    predictions = model.predict(img_array)
    
    pred_class = class_names[np.argmax(predictions[0])]
    confidence = round(100*(np.max(predictions[0])), 2)
    
    return pred_class, confidence
```

```python
plt.figure(figsize=(16, 16))
for imgs, labels in testing_dataset.take(1):  
    for i in range(9):
        plt.subplot(3,3,i+1)
        plt.imshow(imgs[i].numpy().astype('uint8'))
        pred_class, conf = predict_confidence(model, imgs[i])
        true_class = class_names[labels[i]]
        
        plt.title(f"True label: {true_class}.\nPredicted label: {pred_class}.\nConfidence: {conf}%")
        plt.axis('off')
```
